```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()
```

```python
tip = sns.load_dataset("tips")
print(type(tip))
print(tip.head())
```

```python
sns.set_theme()
sns.relplot(
    data = tip,
    x = "total_bill",
    y="tip",
    hue = "smoker",
    size= "size",
    style = "smoker"
)
```

```python
sns.scatterplot(
    data = tip,
    x = "total_bill",
    y="tip",
    hue = "time",
   
    
)
```

```python
flights = sns.load_dataset("flights")

print(flights.head())
```

```python
sns.lineplot(
    data = flights,
    x = "year",
    y="passengers",
    hue = "month",
   
    
)
```

```python
sns.barplot(
    data = tip,
    x = "day",
    y="tip",
    hue = "sex",
   
    
)
```

```python
sns.boxplot(
    data = tip,
    x = "day",
    y="tip"
   
    
)
```

```python
penguins = sns.load_dataset("penguins")

print(penguins.head())
```

```python
sns.histplot(
    data = penguins ,
    x = "body_mass_g",
    bins = 30
   
    
)
```

```python
flights_pivot = flights.pivot(index="month",columns ="year",values ="passengers")
sns.heatmap(
    data = flights_pivot,
    cmap = "coolwarm",
    annot = True,
    fmt = 'd'

    
)
plt.tight_layout()
```
