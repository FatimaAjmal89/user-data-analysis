```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M
```

```python
plt.plot( years,oscar_revenue)
plt.title("Oscar Movies Revenue")
plt.xlabel("Movie year")
plt.ylabel("oscar_revenue in $M")
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.plot( years,oscar_revenue,color = "pink", marker = ">" ,label = "Oscar movies")
plt.plot( years,non_oscar_revenue, marker = ">" ,linestyle = "--", label = "Non Oscar movies")
plt.title("Oscar Movies Revenue VS Non Oscar Movies")
plt.xlabel("Movie year")
plt.ylabel("oscar_revenue in $M")

plt.legend()
```
