```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()
```

```python
tip = sns.load_dataset("tips")
print(type(tip))
print(tip.head())
```

```python
sns.set_theme()
sns.relplot(
    data = tip,
    x = "total_bill",
    y="tip",
    hue = "smoker",
    size= "size",
    style = "smoker"
)
```

```python
sns.scatterplot(
    data = tip,
    x = "total_bill",
    y="tip",
    hue = "time",
   
    
)
```
