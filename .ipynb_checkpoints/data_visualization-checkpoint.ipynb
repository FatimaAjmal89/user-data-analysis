```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M
```

```python
plt.plot( years,oscar_revenue)
plt.title("Oscar Movies Revenue")
plt.xlabel("Movie year")
plt.ylabel("oscar_revenue in $M")
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.plot( years,oscar_revenue,color = "pink", marker = ">" ,label = "Oscar movies")
plt.plot( years,non_oscar_revenue, marker = ">" ,linestyle = "--", label = "Non Oscar movies")
plt.title("Oscar Movies Revenue VS Non Oscar Movies")
plt.xlabel("Movie year")
plt.ylabel("oscar_revenue in $M")

plt.legend()
```

```python
oscar_movies = [ "The Dark Knight","The Hurt Locker","The King's Speech", "The Artist", "Argo"]
years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232]  # in $M
non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M
x = np.arange(len(years))
width= 0.4
plt.bar(x+width/2, revenue,width,label = "oscar movies")
plt.bar( x-width/2,non_oscar_revenue,width,label = "non oscar movies")



plt.title("Oscar vs Non Oscar Movies")
plt.xlabel("years")
plt.ylabel("revenue")
plt.legend()
```

```python
people = ["Person A", "Person B", "Person C", "Person D", "Person E", 
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

plt.scatter(age , bp , s = bp ,cmap = "Wistia", c = bp )
plt.title("BP by age")
plt.ylabel("BP")
plt.xlabel("PEOPLE")
plt.grid()
plt.colorbar(label  = "BP")

for i in range(len(people)):
    plt.annotate(people[i], xy =(age[i], bp [i]) , xytext = (age[i]+1, bp  [i]+1))

plt.xlim (min(age) , max(age)+10)
plt.ylim (min(bp) , max(bp)+10)
```

```python
# Multiple Plots

cities = ["City A", "City B", "City C", "City D", "City E"]

# Winter: Temperature (°C) vs Humidity (%)[cite: 2]
winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]
# Summer
summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]

plt.scatter(winter_temp ,winter_humidity, label = "winter" )
plt.scatter(summer_temp ,summer_humidity, label = "hot" )

plt.title("Humidity")
plt.ylabel("humidity")
plt.xlabel("Temp")
plt.legend()
```

```python
# Pie Charts

expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]
color = ['teal','blue','red','pink','purple']
plt.pie(amounts , labels = expenses , autopct = "%1.1f%%" , colors= color
       , wedgeprops = {"edgecolor": "black"}
        ,shadow = True
       )
```

```python
# multiple datasets
legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

plt.hist(fraud_transactions, color = "red",bins = 20,alpha =0.5 , )
plt.hist(legit_transactions, color = "green", alpha =0.5, bins = 20 )

plt.axvline(
    x = 300,
    label = "profit",
    color = "blue"
    
)

plt.title("Transaction")
plt.ylabel("frequency")
plt.xlabel("Amount in $")
plt.legend()
```
